```python
import pandas as pd

pd.set_option("display.max_colwidth", 150)

caminho = "../data/raw/"

# lendo os datasets, vamos focar no reviews
# aproveitando para ler todos os datasets caso sobre tempo para o desafio
order = pd.read_csv(caminho + "olist_orders_dataset.csv")
reviews = pd.read_csv(caminho + "olist_order_reviews_dataset.csv")
items = pd.read_csv(caminho + "olist_order_items_dataset.csv")
products = pd.read_csv(caminho + "olist_products_dataset.csv")
customers = pd.read_csv(caminho + "olist_customers_dataset.csv")

print("orders shape: ", order.shape)
print("reviews shape: ", reviews.shape)
print("items shape: ", items.shape)
print("products shape: ", products.shape)
print("customers shape: ", customers.shape)
```

```python
# verificando dataset de reviews
reviews.info()
reviews.head(10)
```

```python
# analisando dataset
# total de nulos na coluna de review_comment_title e review_comment_message
total_avaliacoes = len(reviews)

print("Total de avaliações: ", total_avaliacoes)

total_nulos_title =reviews["review_comment_title"].isnull().sum()
print("Total de avaliações sem título: ", total_nulos_title)

total_nulos_message =reviews["review_comment_message"].isnull().sum()
print("Total de avaliações sem mensagem: ", total_nulos_message)

print("% mensagem nula:", round(reviews["review_comment_message"].isnull().mean() * 100, 1), "%")
```

```python
# analisando a distribuição das notas

# dataset somente com os reviews que possuem texto
com_texto = reviews[reviews["review_comment_message"].notnull()].copy()

print("Total de avaliações com texto: ", len(com_texto))
print("Distribuição de todas as notas")
print(reviews['review_score'].value_counts(normalize=True).sort_index() * 100)
print()
print("Distribuição de notas de quem comentou")
print(com_texto['review_score'].value_counts(normalize=True).sort_index() * 100)
```

as distribuicoes dos comentario por nota, tanto do dataset inteiro como do que tem somente comentarios preenchidos, parecem ter a mesma tendencia, observamos que as pessoas que mais comentam atribuem notas altas como 4 e 5, mas tambem as que atribuem notas ruins como a nota 1.

```python
# analisando tamanho dos comentarios
com_texto["tamanho_caracteres"] = com_texto["review_comment_message"].str.len()
com_texto["tamanho_palavras"] = com_texto["review_comment_message"].str.split().str.len()

print(com_texto[["tamanho_caracteres", "tamanho_palavras"]].describe())
```

 As medias sao aproximadamente 69 caracteres e 12 palavras com no maximo 208 caracteres e 45 palavras por comentario.

 Parecem ser comentarios breves, 75% ficam com 16 palavras.

```python
# vendo exemplos
print("Exemplo curto")
print(com_texto.sort_values("tamanho_palavras").iloc[0]["review_comment_message"])
print()
print(" Exemplo mediano")
meio = len(com_texto) // 2
print(com_texto.sort_values("tamanho_palavras").iloc[meio]["review_comment_message"])
print()
print(" Exemplo longo")
print(com_texto.sort_values("tamanho_palavras", ascending=False).iloc[0]["review_comment_message"])
```

```python
com_texto.sort_values("tamanho_palavras", ascending=True)
```

Ao analisar mais a fundo e visualizar alguns exemplos, notamos que existem strings vazias que n estao marcadas como nulas, por isso o describe de quantidade de palavras retornou 0 como menor comentario.

```python
# quantidade de comentarios quando retiramos os nulos
com_texto.shape
```

```python
# de fato quantos comentarios temos retirando tmb as strings vazias
com_texto["texto_stripped"] = com_texto["review_comment_message"].str.strip()

vazios_disfarcados = (com_texto["texto_stripped"] == "").sum()
print("Comentários 'string vazias':", vazios_disfarcados)
print("% sobre quem 'tinha texto':", round(vazios_disfarcados / len(com_texto) * 100, 2), "%")
print()
com_texto_real = com_texto[com_texto["texto_stripped"] != ""].copy()
print("Total de avaliações com texto REAL, após limpeza:", len(com_texto_real))
```

```python
com_texto_real["tamanho_palavras_real"] = com_texto_real["texto_stripped"].str.split().str.len()

print(com_texto_real["tamanho_palavras_real"].describe())
print()
print("Comentários com 1 palavra só:")
print(com_texto_real[com_texto_real["tamanho_palavras_real"] == 1]["texto_stripped"].head(10).tolist())
```

O volume de comentarios com string vazias e bem baixo de 0.07%, mesmo assim vale retirar tmb estes ruidos.

A base fica entao com 40950 registros.

Tambem verificamos os comentarios com apenas uma palavra, parecem contribuir para uma tendencia positiva ou negativa.
