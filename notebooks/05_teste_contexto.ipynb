```python
import sys
sys.path.append("..")

from src.embeddings import NILCEmbeddings
from src.vectorstore import carregar_colecao
from src.llm import criar_llm
from src.retrieval import recuperar_multiconsulta
from src.context import construir_contexto, extrair_citacoes, validar_citacoes
```

```python
store = carregar_colecao(NILCEmbeddings())
llm = criar_llm()

pergunta = "Quais são os principais problemas relatados pelos clientes?"
consultas, evidencias = recuperar_multiconsulta(store, pergunta, llm)
contexto = construir_contexto(evidencias)

print(contexto.texto)
print(f"\nTamanho: {len(contexto.texto)} caracteres (~{len(contexto.texto) // 4} tokens)")

print("\nMapa de rastreabilidade (E# -> review_ids):")
for id_ev, ev in contexto.evidencias.items():
    print(f"  {id_ev}: {ev.review_ids}")

# Teste da verificação de citações com uma resposta fictícia
resposta_teste = "Há atrasos [E1][E6], produtos com defeito [E2, E7] e algo inventado [E99]."
print("\nCitações extraídas:", extrair_citacoes(resposta_teste))
print("Válidas / inexistentes:", validar_citacoes(resposta_teste, contexto))
```
